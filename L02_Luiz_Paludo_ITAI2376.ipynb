**Introduction**

We read the introduction and understood that the main goal of this lab is to learn how deep learning and pre-trained models work using a no-code approach with Google Colab and VGG16.

```python
import tensorflow as tf
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np

model = VGG16(weights='imagenet')
model.summary()
```

**Image Preprocessing**

This function gets the image ready to be used by VGG16. It changes the image size to 224 × 224, converts it into an array, and applies the preprocessing needed by the model before making a prediction.


```python
from tensorflow.keras.utils import load_img, img_to_array

def prepare_image(img_path):
    img = load_img(img_path, target_size=(224, 224))
    x = img_to_array(img)
    x = np.expand_dims(x, axis=0)
    x = preprocess_input(x)
    return x
```

**Making a Prediction**

Here, I use the prepared image as input for VGG16. The model analyzes the image and gives the three most likely predictions with their confidence scores.


```python

img_path = '/content/Steer.jpg'
image_data = prepare_image(img_path)
preds = model.predict(image_data)

results = decode_predictions(preds, top=3)[0]
for i, (imagenet_id, label, score) in enumerate(results):
    print(f"{i+1}. {label}: {score*100:.2f}%")
```

**Uploading an Image**

This cell lets me upload an image from my computer. After the upload, the image is prepared and sent to VGG16, which then shows the three predictions with the highest confidence.


```python
from google.colab import files
from tensorflow.keras.utils import load_img, img_to_array
import numpy as np

uploaded = files.upload()

for fn in uploaded.keys():
    img_path = fn
    image_data = prepare_image(img_path)
    preds = model.predict(image_data)

    results = decode_predictions(preds, top=3)[0]
    print(f"\n--- Results for {fn} ---")
    for i, (imagenet_id, label, score) in enumerate(results):
        print(f"{i+1}. {label}: {score*100:.2f}%")
```
